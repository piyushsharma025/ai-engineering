async and await

async tells Python that a function can perform asynchronous operations.

await tell Python:
    Pause this coroutine until this asynchronous operation is ready but allow the event loop to do other work.
    await can only be used inside an async function.

```python
import asyncio

async def hello():
    await asyncio.sleep(2)
    print("Hello!")

print("Before")

#await is required to call the corutine functions
await hello()   #do this in notebook

# asyncio.run(hello())  do this in .py line

print("After")
```

Concurrency with asyncio.gather()

without asyncio.gather(), we waited for taks a to finish before starting task b.

but if the tasks are independent, we can run them concurrently. 

asycio.gather() schedules multiple coroutines and waits for all them to complete.

This is particularly useful for I/O-bound operations such as :
- API calls
- DB queries
- LLM requests
- File/Network operations

```python
import asyncio

async def task_a():
    print("Task A started")
    await asyncio.sleep(2)
    print("Task A finished")

async def task_b():
    print("Task B Started")
    await asyncio.sleep(1)
    print("Task B is finished")

async def main():
    result = await asyncio.gather(task_a(), task_b())

    print(result)

await main()
```

Event Loop-

The event loop is the engine behind asyncio.

It keeps track of asynchronous tasks and resumes them when they are ready to make progress.

When one task is waiting for I/O, the event loop can work on another task. You normally don't need to manually managet the event loop.

For Standalone Python program, aysncio.run() is the common entry point.


```python
import asyncio

async def generate_number():

    for number in range(5):
        await asyncio.sleep(1)
        yield number

async def main():

    async for number in generate_number():
        print(number)

await main()
```

```python
import httpx
import asyncio

async def fetch_post():

    async with httpx.AsyncClient() as client:

        response = await client.get("https://jsonplaceholder.typicode.com/posts/1")

        return response.json()

async def main():
    post = await fetch_post()#This creates a coroutine object. (To actually run the coroutine, it needs to be executed by an event loop.)
    print(post)

await main()
```
