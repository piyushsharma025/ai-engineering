```python
class Dog:

    species = "Dog"

    def __init__(self, name: str, breed: str):
        self.name = name
        self.breed = breed

    def bark(self):
        return f"{self.name} says woof!"

    @classmethod
    def walk(cls):
        print(f"{cls.species} is walking")


Dog.walk()
leo = Dog(name="Leo", breed="Labra")
print(leo.bark())
print(leo.breed)
        
```

```python
class User:

    def __init__(self, first, last):
        self.first = first
        self.last = last

    @property
    def full_name(self):
        return f"{self.first} {self.last}"

piyush = User(first = "Piyush", last = "Sharma")
print(piyush.full_name)     
        
```

```python
#Inheritance
class Animal:
    def speak(self):
        print("some sound")

class Dog(Animal):
    def speak(self):
        print("Woof")

dog = Dog()
dog.speak()
```

```python
#Composition
class Tool:
    content: str
    name: str
    description: str

class Agent:

    def __init__(self, model, tool):
        self.model = model
        self.tool = tool

tool = Tool()
agent = Agent(model="GPT-5.1", tool=tool)
```

```python
def add_numbers(args):
    print(args)

add_numbers(1, 2, 3) # will not work because function only accepts one argument
```

```python
def add_numbers(*args): #This is how you can give as many parameters(*args)
    print(args)
    print(type(args))#Tuple

add_numbers(1, 2, 3, "Piyush", True)
```

```python
def print_profile(**kwargs):
    print(kwargs)
    print(type(kwargs))#dict

print_profile(name = "Piyush", age = 26, role="SE")
```

```python
#Unpacking a dictionary
def create_profile(name, role, experience):
    print(name)
    print(role)
    print(experience)

profile = {
    "name" : "Shivam",
    "role": "SSE",
    "experience": 4
}
create_profile(name="Piyush", role = "SSE", experience=5)
create_profile(**profile)
```
