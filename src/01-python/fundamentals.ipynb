```python
class APIValidationError(Exception): #It's extending from Excpetion Class
    pass  #don't want to give defination

def process_api_payload(payload : dict):
    if "prompt" not in payload:
        raise APIValidationError("Missing required 'prompt' key in payload.")
    return f"Processing prompt: {payload['prompt']}"

def execute_pipeline(payload : dict):
    try:
        print("starting pipeline execution..")
        result = process_api_payload(payload)
    except APIValidationError as err:
        print(f"Caught specific api validation error : {err}")
    except Exception as err:
        print(f"caught unexpected generic error: {err}")
    else:
        print(f"pipeline succeded: {result}")
    finally:
        print("claenup..")

print("--executing with invalid payload---")
execute_pipeline({})

print("---executing with valid payload--")
execute_pipeline({"prompt" : "Summarize this text"})
```
